```python
from ucimlrepo import fetch_ucirepo

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    roc_auc_score,
    RocCurveDisplay,
    precision_recall_curve, 
    PrecisionRecallDisplay
)
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


# =========================
# LOAD DATASET
# =========================

# fetch dataset 
fertility = fetch_ucirepo(id=244) 
  
# data (as pandas dataframes) 
X = fertility.data.features 
y = fertility.data.targets 
  
# metadata 
# print(fertility.metadata) 
  
# variable information 
print(fertility.variables) 

# trasformiamo in DataFrame unico (opzionale)
df = pd.concat([X, y], axis=1)

print(df.describe())
print(df.shape)
print(df.head(15))
print(df["diagnosis"].dtype)
print(df["diagnosis"].value_counts())

# =========================
# TARGET ENCODING
# =========================

# Trasforma target in numerico: 0 = Normal (N), 1 = Altered (O) - one hot encoding
df['diagnosis'] = df['diagnosis'].map({'N': 0, 'O': 1})
```

```python
# =========================
# EDA
# =========================

min_age = 18
max_age = 36
 
df['age_real'] = df['age'] * (max_age - min_age) + min_age
 
# Conta quante persone ci sono per ogni età
age_counts = df['age_real'].value_counts().sort_index()
 
# Grafico a barre
plt.bar(age_counts.index, age_counts.values)
plt.xlabel("Età (anni)")
plt.ylabel("Numero di persone")
plt.title("Distribuzione delle età")
plt.show()
 
plt.boxplot(df['alcohol'], vert=False)
plt.xlabel("Alcol (anni)")
plt.title("Boxplot Alcol")
plt.show()
```

```python
# =========================
# FEATURE SELECTION
# =========================

# Lista delle features da analizzare
features = ["alcohol", "smoking", "hrs_sitting"]

# Ciclo su ciascuna feature per eseguire:
# 1) Tabella di contingenza
# 2) Visualizzazione grafica delle proporzioni
for col in features:
    print(f"\n===== {col.upper()} =====")

    # 1) Tabella di contingenza (frequenze assolute)
    # Mostra il numero di osservazioni per ogni combinazione
    # tra la feature corrente e la variabile target (diagnosis)
    table = pd.crosstab(df[col], df["diagnosis"])
    print(table)

    # Tabella normalizzata per riga (opzionale)
    table_percent = pd.crosstab(
        df[col], df["diagnosis"], normalize="index"
    )
    
    table_percent.plot(kind="bar", stacked=True)
    plt.title(f"{col} vs Diagnosis")
    plt.xlabel(col)
    plt.ylabel("Proportion")
    plt.tight_layout()
    plt.show()
```

```python
# =========================
# STRATIFIED TRAIN / TEST SPLIT
# =========================

# Definire le feature da usare tutte insieme - Multi-feature
#1) Preparazione dei dati
X = df[['alcohol', 'smoking', 'hrs_sitting']]
y = df['diagnosis']

# --- Train/Test split - Training 80%/Test 20% ---
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

# =========================
# LOGISTIC REGRESSION
# =========================
log_model = LogisticRegression(class_weight='balanced')
log_model.fit(X_train, y_train)

# --- Predizione ---
y_pred_log = log_model.predict(X_test)
y_proba_log = log_model.predict_proba(X_test)[:, 1]

print("\n===== LOGISTIC REGRESSION =====")
print(classification_report(y_test, y_pred_log))

roc_auc_log = roc_auc_score(y_test, y_proba_log)
print("ROC-AUC:", roc_auc_log)

# Confusion Matrix
cm_log = confusion_matrix(y_test, y_pred_log)
ConfusionMatrixDisplay(cm_log).plot()
plt.title("Confusion Matrix - Logistic")
plt.show()

# ROC Curve
RocCurveDisplay.from_predictions(y_test, y_proba_log)
plt.title("ROC Curve - Logistic")
plt.show()

# Precision-Recall Curve
precision, recall, thresholds = precision_recall_curve(y_test, y_proba_log) # Calcolo i valori di precisione e recall per diverse soglie di classificazione
PrecisionRecallDisplay(precision=precision, recall=recall).plot()           # thresholds --> valori di soglia corrispondenti ai punti di precision-recall
plt.title("Precision-Recall Curve - Logistic Regression")
plt.show()

# Cross Validation
cv_scores_log = cross_val_score(log_model, X, y, cv=5, scoring='roc_auc')
print("Cross-Validation ROC-AUC (mean):", cv_scores_log.mean())
```

```python
# ==============================
# INTERPRETAZIONE COEFFICIENTI
# ==============================
coeff_df = pd.DataFrame({
    'Feature': features,
    'Coefficient': log_model.coef_[0],
    'Odds Ratio': np.exp(log_model.coef_[0])
})

# odds ratio: quanto aumenta (o diminuisce) la probabilità di avere diagnosis=1 se la feature aumenta di 1 unità
# log_model.coef_[0] = array dei coefficienti β della regressione logistica.

# center = 1 --> odds ratio = 1, significa nessun effetto.
# Valori > 1 --> aumento rischio
# Valori < 1 --> diminuzione rischio

print("\nOdds Ratio:")
print(coeff_df)

sns.heatmap(
    coeff_df.set_index('Feature')[['Odds Ratio']],
    annot=True,
    cmap="coolwarm",
    center=1
)
plt.title("Odds Ratio - Logistic Regression")
plt.show()
```

```python
# =========================
# DECISION TREE
# =========================

# max_depth=3 per avere regole semplici e interpretabili
dt_model = DecisionTreeClassifier(
    max_depth=3, # profondità massima dell'albero
    min_samples_leaf=5, # almeno 5 campioni per foglia (regolarizzazione)
    class_weight='balanced',
    random_state=42)

dt_model.fit(X_train, y_train)

# --- Predizione su test set ---
y_pred_dt = dt_model.predict(X_test)
y_proba_dt = dt_model.predict_proba(X_test)[:, 1]

print("\n===== DECISION TREE =====")
print(classification_report(y_test, y_pred_dt))

roc_auc_dt = roc_auc_score(y_test, y_proba_dt)
print(f"ROC-AUC: {roc_auc_dt}")

# Confusion Matrix
cm_dt = confusion_matrix(y_test, y_pred_dt)
ConfusionMatrixDisplay(cm_dt).plot()
plt.title("Confusion Matrix - Decision Tree") 
plt.show()

# ROC Curve
RocCurveDisplay.from_predictions(y_test, y_proba_dt)
plt.title("ROC Curve - Decision Tree")
plt.show()

# Precision-Recall Curve
precision, recall, thresholds = precision_recall_curve(y_test, y_proba_dt)
PrecisionRecallDisplay(precision=precision, recall=recall).plot()           
plt.title("Precision-Recall Curve - Decision Tree")
plt.show()

# Cross Validation
cv_scores_dt = cross_val_score(
    dt_model,
    X,
    y,
    cv=5,
    scoring='roc_auc'
)
print(f"Cross-Validation ROC-AUC (mean): {cv_scores_dt.mean()}")

# Feature Importance
importances_dt = pd.Series(dt_model.feature_importances_, index=features)
importances_dt.sort_values().plot(kind='barh')
plt.title("Feature Importance - Decision Tree")
plt.show()

# La feature importance misura quanto ogni variabile contribuisce alle decisioni del modello. In pratica risponde a:
# "Quale variabile è più utile per distinguere Normal da Altered?"

# =========================
# VISUALIZZAZIONE ALBERO
# =========================
plt.figure(figsize=(15, 8))
plot_tree(
    dt_model,
    feature_names=features, 
    class_names=['Normal', 'Altered'],
    filled=True, # colora i nodi in base alla classe
    rounded=True # arrotonda gli angoli dei box
)
plt.title("Decision Tree (max_depth=3)")
plt.show()
y_pred = dt_model.predict(X_test)
print(f"Predizioni su test set: {y_pred}")
```

```python
# =================
# RANDOM FOREST
# =================

rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=4,
    class_weight='balanced',
    random_state=42
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
y_proba_rf = rf_model.predict_proba(X_test)[:, 1]

print("\n===== RANDOM FOREST =====")
print(classification_report(y_test, y_pred_rf))

roc_auc_rf = roc_auc_score(y_test, y_proba_rf)
print("ROC-AUC:", roc_auc_rf)

# Confusion Matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf)
plt.title("Confusion Matrix - Random Forest")
plt.show()

# ROC Curve
RocCurveDisplay.from_predictions(y_test, y_proba_rf)
plt.title("ROC Curve - Random Forest")
plt.show()

# Precision-Recall Curve
precision, recall, thresholds = precision_recall_curve(y_test, y_proba_rf)
PrecisionRecallDisplay(precision=precision, recall=recall).plot()
plt.title("Precision-Recall Curve - Random Forest")
plt.show()

# Cross Validation
cv_scores_rf = cross_val_score(rf_model, X, y, cv=5, scoring='roc_auc')
print("Cross-Validation ROC-AUC (mean):", cv_scores_rf.mean())

# Feature Importance
importances_rf = pd.Series(rf_model.feature_importances_, index=features)
importances_rf.sort_values().plot(kind='barh')
plt.title("Feature Importance - Random Forest")
plt.show()
```

```python
# =====================
# GRADIENT BOOSTING
# =====================

gb_model = GradientBoostingClassifier(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

gb_model.fit(X_train, y_train)

y_pred_gb = gb_model.predict(X_test)
y_proba_gb = gb_model.predict_proba(X_test)[:, 1]

print("\n===== GRADIENT BOOSTING =====")
print(classification_report(y_test, y_pred_gb, zero_division=0))

roc_auc_gb = roc_auc_score(y_test, y_proba_gb)
print("ROC-AUC:", roc_auc_gb)

# Confusion Matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_gb)
plt.title("Confusion Matrix - Gradient Boosting")
plt.show()

# ROC Curve
RocCurveDisplay.from_predictions(y_test, y_proba_gb)
plt.title("ROC Curve - Gradient Boosting")
plt.show()

# Precision-Recall Curve
precision, recall, thresholds = precision_recall_curve(y_test, y_proba_gb)
PrecisionRecallDisplay(precision=precision, recall=recall).plot()
plt.title("Precision-Recall Curve - Gradient Boosting")
plt.show()

# Cross Validation
cv_scores_gb = cross_val_score(gb_model, X, y, cv=5, scoring='roc_auc')
print("Cross-Validation ROC-AUC (mean):", cv_scores_gb.mean())

# Feature Importance
importances_gb = pd.Series(gb_model.feature_importances_, index=features)
importances_gb.sort_values().plot(kind='barh')
plt.title("Feature Importance - Gradient Boosting")
plt.show()
```
